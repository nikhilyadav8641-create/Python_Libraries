```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

from google.colab import files

uploaded = files.upload()
filename=list(uploaded.keys())[0]
df=pd.read_csv(filename)
# df.head()
df.describe()#descriptive statistics ,it will calculate avg value , stantard deviatiation ,min value ,max value , 25%, 50% , 75%
# If you meant to access the first element of the first column by position, use .iloc:
df.iloc[0,0]==50
df.head(4)# To print first 4 rows
df.tail(2)# To print last 2 rows
df = df.sample(frac=1)# it willm show whole data set and it will shuffle the data set

# finding that on which feature the target depend the most
correlation = df.corr(numeric_only=True)["SalePrice"].sort_values(ascending=False)

print(correlation)

# duplicate of some field
df["SalePrice_copy"] = df["SalePrice"]

#reverse the fields
df["SalePrice"] = df["SalePrice"].iloc[::-1].values
print(df["SalePrice"].iloc[::-1])

# seprating the target features

y = df["SalePrice"]

X = df.drop("SalePrice", axis=1)

# selecting the numerical column
X_numerical = X.select_dtypes(include=np.number)

X_numerical = X_numerical.fillna(X_numerical.mean())
print(X_numerical)

# spliting the data
X_train, X_test, y_train, y_test = train_test_split(
    X_numerical,
    y,
    test_size=0.3,
    random_state=50
)

```

```python
# training of model
model = LinearRegression()

model.fit(X_train, y_train)
y_pred = model.predict(X_test)
# ploting the data points via matplotlib
plt.figure(figsize=(10, 6))

sns.scatterplot(
    x=y_test,
    y=y_pred
)

plt.xlabel("Actual SalePrice")
plt.ylabel("Predicted SalePrice")

plt.title("Actual vs. Predicted SalePrice")


# Diagonal reference line
plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    linestyle="--"
)

plt.grid(True)

plt.show()
```
